```python
%sql
--  Drop the table if it exist for demonstration purpose 
DROP TABLE IF EXISTS current_employees_ctas;

--  Create Table
CREATE TABLE current_employees_ctas as 
select * from read_files
("/Volumes/dbacademy/0002lakeflowconnectingestion/myfiles/1782230321-CustomerMaster.csv",
    format=>"csv",
    header=>true,
    inferSchema=>true);

-- Show Table
show tables;
```
