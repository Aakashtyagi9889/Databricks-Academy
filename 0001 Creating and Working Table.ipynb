```python
%sql
--  Create table 
DROP TABLE IF EXISTS current_employees;

--  Create a Delta Table using csv Files 
CREATE TABLE current_employees AS
SELECT * from read_files(
    '/Volumes/dbacademy/default/raw_data',
    format=> 'csv',
    header=>true,
    inferSchema=>true
);

-- Display Table 
Select * from current_employees;
```

### PySpark

```python
# Read the csv and create a spark Dataframe
df = (spark.read.format("csv")\
    .option("header","true")\
    .option("inferSchema","true")\
    .load("/Volumes/dbacademy/default/raw_data/rides/build_data_pipeline_demo/"))
df.display()
```

```python
#  Create 
df.write.mode("overwrite").format("delta").saveAsTable("dbacademy.default.current_employees_py2")
```

Read the Delta table using python


```python
spark.read.table("dbacademy.default.current_employees_py2").display()
```
